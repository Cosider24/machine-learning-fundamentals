#Часть 2

2.1 Добавьте новый столбец is_large_flower, который будет содержать True, если сумма sepal.length + petal.length больше 10, и False — в противном случае. Этот столбец поможет быстро определить, какие цветы считаются крупными.

```python
df['is_large_flower'] = (df['sepal.length'] + df['petal.length']) > 10
```

2.2 Добавьте еще два столбца:

variety.upper — название вида большими буквами.

variety.short — первые 3 буквы вида

```python
df['variety.upper'] = df['variety'].str.upper()
df['variety.short'] = df['variety'].str[:3]
```

2.3 Создание вспомогательного DataFrame

Создайте отдельный DataFrame df_aux с колонками variety и sepal.length (копия df).
Добавьте столбец sepal_category, который будет 'long', если sepal.length > 5.8, и 'short' иначе.

```python
df_aux = df[['variety', 'sepal.length']].copy()
df_aux['sepal_category'] = ['long' if x > 5.8 else 'short' for x in df_aux['sepal.length']]
```

2.4 Объединение таблиц

Объедините df и df_aux с помощью merge по столбцу variety (inner join).

Объедините df и df_aux с помощью concat по столбцам (axis=1).

Объясните разницу между merge и concat.

```python
df_merged = pd.merge(df, df_aux, on='variety', how='inner')
df_concat = pd.concat([df, df_aux], axis=1)
```
